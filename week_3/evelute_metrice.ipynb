```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Tiny dataset
X = np.array([1, 2, 3, 4, 5, 6, 7, 8], dtype=float)
y = np.array([20, 30, 40, 50, 60, 70, 80, 85], dtype=float)
```

```python
X = X.reshape(-1, 1)
```

```python
X,y
```

```python
from sklearn.linear_model import LinearRegression
model = LinearRegression()
model.fit(X, y)
```

```python
def compute_cost(X, y, m, c):
    n = len(y)
    y_pred = m * X + c  # predictions
    error = y - y_pred
    cost = (1 / n) * np.sum(error**2)
    return cost
```

```python
print(compute_cost(X, y, m=0, c=0))  # bad line, huge cost
print(compute_cost(X, y, m=10, c=10))  # better line, smaller cost
```

```python
print("M: ", model.coef_[0])
print("C: ", model.intercept_)
```

```python
plt.scatter(X, y, color='blue', label='Data Points')
predicted_y = model.predict(X)
plt.plot(X, predicted_y, color='red', label='Regression Line')
plt.xlabel('X')
plt.ylabel('y')
plt.title('Linear Regression Fit')
plt.legend()
plt.grid()
plt.show()
```

```python
new_hours = np.array([[7]])
prediction = model.predict(new_hours)
print(f"Predicted score for 7 hours of study: {prediction[0]:.2f}")
```

```python
from sklearn.metrics import mean_squared_error
```

```python
pred_marks = model.predict(X)
mse = mean_squared_error(y, pred_marks)
print(f"Mean Squared Error: {mse:.2f}")
```

```python
plt.scatter(X, y, color="blue", label="Data Points")
predicted_y = model.predict(X)
plt.plot(X, predicted_y, color="red", label="Regression Line")
plt.xlabel("X")
plt.ylabel("y")
plt.title("Linear Regression Fit")
plt.legend()
plt.grid()
plt.show()
```

```python
from sklearn.linear_model import SGDRegressor
```

```python
sgd_reg = SGDRegressor(
    max_iter=1000,
    tol=1e-3,
    learning_rate='invscaling',
    eta0=0.01,
    random_state=42
)
```

```python
sgd_reg.fit(X, y)
print("M (SGD): ", sgd_reg.coef_[0])
print("C (SGD): ", sgd_reg.intercept_[0])
```

```python
pred_sgd = sgd_reg.predict(X)
mse_sgd = mean_squared_error(y, pred_sgd)
print(f"Mean Squared Error (SGD): {mse_sgd:.2f}")
```

```python
plt.scatter(X, y, color="blue", label="Data Points")
predicted_y = model.predict(X)
plt.plot(X, pred_marks, color="red", label="Regression Line", linestyle='--')
plt.plot(X, pred_sgd, color="green", label="SGD Regression Line", linestyle='-.')
plt.xlabel("X")
plt.ylabel("y")
plt.title("Linear Regression Fit VS SGD Regression Fit")
plt.legend()
plt.grid()
plt.show()
```

```python
from sklearn.metrics import mean_absolute_error, r2_score
```

```python
mae = mean_absolute_error(y, pred_marks)
mse = mean_squared_error(y, pred_marks)
r2 = r2_score(y, pred_marks)
rmse = np.sqrt(mse)
print(f"Mean Absolute Error: {mae:.2f}")
print(f"Mean Squared Error: {mse:.2f}") 
print(f"Root Mean Squared Error: {rmse:.2f}")
print(f"R^2 Score: {r2:.2f}")
```
