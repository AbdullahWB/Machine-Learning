## Q2. Bayes and Probability in ML (10 marks)

**Given:**
Prevalence (P(D)=p). Sensitivity (=\text{Se}=0.90). Specificity (=\text{Sp}=0.92).
For a **positive test**, PPV (= P(D \mid +)) is:
[
\text{PPV} ;=; \frac{\text{Se}\cdot p}{\text{Se}\cdot p + (1-\text{Sp})\cdot(1-p)}.
]

**(a) p = 1% = 0.01**
[
\text{PPV}=\frac{0.90\cdot 0.01}{0.90\cdot0.01 + 0.08\cdot0.99}
= \frac{0.009}{0.009 + 0.0792} \approx 0.102 ; (10.2%).
]

**(b) p = 20% = 0.20**
[
\text{PPV}=\frac{0.90\cdot 0.20}{0.90\cdot0.20 + 0.08\cdot0.80}
= \frac{0.18}{0.18 + 0.064} \approx 0.7377 ; (73.8%).
]

**Interpretation (3–4 sentences):** When the **prevalence increases**, the **prior** probability of disease is higher, so a positive test is **more likely** to be a true positive. Thus **PPV jumps** from ~10% at 1% prevalence to ~74% at 20% prevalence—even with the same sensitivity/specificity. This is the diagnostic-testing application of **Bayes’ rule**: posterior (PPV) depends critically on both test properties and **prevalence**.
